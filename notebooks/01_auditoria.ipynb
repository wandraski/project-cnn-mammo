## 01 · Auditoria de dados


Sanidade: pacientes duplicados, vazamento entre splits, distribuição por vista/lateralidade.


```python
from pathlib import Path
from project_cnn_mammo.data import read_manifest, split_plan
manifesto = Path('data/manifests/cbisddsm.csv')
if not manifesto.exists():
    raise SystemExit('rode 00_ingestao primeiro (ou anexe o CBIS-DDSM)')
entries = read_manifest(manifesto)
plan = split_plan(entries)
print(plan.counts)
assert plan.leakage_free, 'vazamento detectado'
```

```python
from collections import Counter
print(Counter((e.view, e.laterality) for e in entries))
print(Counter(e.label for e in entries))
```
