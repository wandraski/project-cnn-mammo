## 02 · Pré-processamento


DICOM → 8 bits, MONOCHROME1, recorte; ablação CLAHE × Z-score.


```python
from project_cnn_mammo.io_dicom import read_dicom, apply_windowing
payload = read_dicom('data/exemplo.dcm')
print({k: payload[k] for k in ('width','height','photometric_interpretation','monochrome1')})
```

```python
from project_cnn_mammo.imaging import preprocess_array
pixels = apply_windowing(payload)
result = preprocess_array(pixels, payload['width'], payload['height'], normalization='clahe_zscore')
print(result['source'], result['tiles'])
```

```python
z = preprocess_array(pixels, payload['width'], payload['height'], normalization='zscore')
print('ablacao registrada:', {'clahe': result['source'], 'zscore': z['source']})
```

## Etapa pré-CNN: aprendizado não supervisionado


```python
from project_cnn_mammo.unsupervised import pre_cnn_flags
features = [[float(v)] for v in pixels[:64]]
flags = pre_cnn_flags(features, method='knn', quantile=0.90)
print('limiar:', round(flags['threshold'], 3), ' anomalias:', flags['anomalies'])
```
